```python
import lightgbm as lgb
import numpy as np
import pandas as pd
from pandas import DataFrame
from sklearn.datasets import load_wine, fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, mean_squared_error
```

### binary example

```python
wine = load_wine()
x, y = DataFrame(wine.data, columns=wine.feature_names), wine.target
```

[train_test_split](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html): splits data arrays into random training and testing subsets.

- test_size: 20% of data, which leaves 80% for training
- random_state: same output if nothing changes, control factor

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

dataset is a class used to create the training data and the testing/validation data

```python
train_data = lgb.Dataset(x_train, label = y_train)  # inde, depend
test_data = lgb.Dataset(x_test, label = y_test)
```

[docs](https://lightgbm.readthedocs.io/en/stable/Parameters.html)

[parameter formats](https://lightgbm.readthedocs.io/en/latest/Parameters.html#parameters-format)

objective: binary (0 1), multiclass (1 2 3 4 5), or regression (continuous)

metric: binary error, RMS error

```python
params = {
    'objective':'binary',  # bc y is binary-classified
    'metric':'binary_error',
    'boosting_type':'gbdt',
    'learning_rate':0.1,
    'verbose':-1,
}
```

training the model for 100 rounds (builds trees one after another)

can describe early stopping criteria as early_stopping_rounds=value

valid sets is an extra dataset that is checked against during training

```python
model_classification = lgb.train(params, train_data, valid_sets = [test_data], num_boost_round=100)
```

```python
y_pred = model_classification.predict(x_test)

y_pred
```

```python
y_pred_binary = [1 if p >= 0.5 else 0 for p in y_pred]

y_pred_binary
```

```python
accuracy_score(y_test, y_pred_binary)
```

### regression example

```python
housing = fetch_california_housing()
```

```python
x = pd.DataFrame(housing.data, columns = housing.feature_names)

y = housing.target  # cont so use regression
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)
```

```python
train_data = lgb.Dataset(x_train, label = y_train)
test_data = lgb.Dataset(x_test, label = y_test)
```

```python
params = {
    'objective':'regression',
    'metric':'rmse',
    'boosting_type':'gbdt',
    'learning_rate':0.05,
    'num_leaves':50,
    'verbose':-1,
}
```

```python
print(type(x_test), x_test.shape if hasattr(x_test, 'shape') else len(x_test))
```

```python
model_regression = lgb.train(params, train_data, valid_sets = [test_data], num_boost_round = 500)
```

```python
y_pred = model_regression.predict(x_test)
```

```python
mean_squared_error(y_test, y_pred)
```
